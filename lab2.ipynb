```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Henry'],
    'department': ['IT', 'HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'HR'],
    'salary': [80000, 60000, 90000, 75000, np.nan, 85000, 70000, 62000],
    'bonus': [5000, 3000, np.nan, 4000, 2000, 6000, 3500, np.nan],
    'hire_date': pd.to_datetime(['2020-01-15', '2019-03-22', '2021-07-01', '2018-11-05',
                                 '2022-02-10', '2020-09-30', '2019-12-12', '2021-05-20'])
}
df = pd.DataFrame(data)

print("Задача 1. Научиться создавать объекты pandas и получать первичную информацию о них")

s = pd.Series([12, 45, 7, 30, 58], index=['q', 'w', 'e', 'r', 't'])
print("\n1.1 Создайте Series из списка чисел с произвольным индексом")
print(s)

print("\n1.2 Создайте DataFrame из словаря")
print(df)

print("\n1.3 Выведите первые и последние строки, информацию о типах данных, статистику")
print("Первые строки:")
print(df.head())
print("\nПоследние строки:")
print(df.tail())
print("\nТипы данных:")
print(df.dtypes)
print("\nСтатистика:")
print(df.describe())

print("\nЗадача 2. Освоить методы доступа к данным и создания новых столбцов")

print("\n2.1 Выберите столбец name и salary для сотрудников из IT")
print(df.query("department == 'IT'")[['name', 'salary']])

print("\n2.2 Используя iloc, выведите первые 3 строки и первые 2 столбца")
print(df.iloc[[0, 1, 2], [0, 1]])

print("\n2.3 Отфильтруйте сотрудников с зарплатой выше 70000")
print(df.loc[df['salary'].gt(70000)])

df['total_comp'] = df['salary'].add(df['bonus'].fillna(0))
print("\n2.4 Создайте новый столбец total_comp = salary + bonus (пропуски в bonus считать нулём)")
print(df[['name', 'salary', 'bonus', 'total_comp']])

del df['hire_date']
print("\n2.5 Удалите столбец hire_date")
print("Остались колонки:", list(df.columns))

print("\nЗадача 3. Научиться обнаруживать и заполнять/удалять пропуски")

print("\n3.1 Подсчитайте количество пропусков в каждом столбце")
print(len(df) - df.count())

means = df.groupby('department')['salary'].mean()
df['salary'] = df['salary'].fillna(df['department'].map(means))
print("\n3.2 Заполните пропуски в salary средним значением по отделу")
print(df[['name', 'department', 'salary']])

df['bonus'] = df['bonus'].replace(np.nan, 0)
print("\n3.3 Заполните пропуски в bonus нулём")
print(df[['name', 'bonus']])

df['total_comp'] = df['salary'] + df['bonus']

df = df.loc[df['salary'].notna()]
print("\n3.4 Удалите строки, где salary всё ещё пропущен (если остались)")
print("Строк осталось:", len(df))

print("\nЗадача 4. Освоить groupby, агрегатные функции и сводные таблицы")

print("\n4.1 Найдите среднюю зарплату по каждому отделу")
print(df.groupby('department')['salary'].mean())

avg = df.groupby('department')['salary'].mean()
mx = df.groupby('department')['bonus'].max()
cnt = df.groupby('department')['name'].count()
stats = pd.concat([avg.rename('avg_salary'), mx.rename('max_bonus'), cnt.rename('employees')], axis=1)
print("\n4.2 Для каждого отдела вычислите: среднюю зарплату, максимальный бонус, количество сотрудников")
print(stats)

print("\n4.3 Создайте сводную таблицу: индекс - отдел, значения - средняя зарплата")
pivot = df.pivot_table(index='department', values='salary', aggfunc='mean')
print(pivot)

print("\n4.4 Отсортируйте отделы по средней зарплате по убыванию")
print(pivot.sort_values('salary', ascending=False))

print("\nЗадача 5. Научиться объединять DataFrame'ы с помощью merge и concat")

managers = pd.DataFrame({
    'department': ['IT', 'HR', 'Finance'],
    'manager': ['Кузнецов', 'Белова', 'Романов']
})
print("\n5.1 Создайте DataFrame managers с отделами и руководителями")
print(managers)

merged = pd.merge(df, managers, on='department', how='left')
print("\n5.2 Объедините df с managers по столбцу department (left join)")
print(merged)

new_emp = pd.DataFrame({'name': ['Ольга'], 'department': ['Finance'], 'salary': [95000],
                        'bonus': [5000], 'total_comp': [100000], 'manager': ['Романов']})
ext = pd.concat([merged, new_emp])
print("\n5.3 Создайте новую строку с данными о сотруднике и добавьте её к df через concat")
print(ext.tail(3))

ext = ext.reset_index(drop=True)
print("\n5.4 Сбросьте индекс после объединения")
print("Индекс теперь:", list(ext.index))

print("\nЗадача 6. Применить изученные операции для ответа на вопросы")

ext['place'] = ext['total_comp'].rank(ascending=False, method='min').astype(int)
print("\n6.1 Найдите топ-3 сотрудников по общей компенсации (total_comp)")
print(ext.loc[ext['place'] <= 3, ['name', 'department', 'total_comp']])

print("\n6.2 Выведите среднюю зарплату и средний бонус по отделам в одном DataFrame")
print(ext.pivot_table(index='department', values=['salary', 'bonus'], aggfunc='mean'))

ext['rich'] = ext['salary'] > 75000
print("\n6.3 Для каждого отдела определите долю сотрудников с зарплатой выше 75000")
print(ext.groupby('department')['rich'].mean())
```
